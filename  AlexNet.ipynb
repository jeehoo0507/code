AlexNet은 2012년 ImageNet 대회에서 우승하며, 깊은 CNN이 대규모 이미지 분류에 효과적이라는 걸 보여준 모델이다.

paper : [link](https://proceedings.neurips.cc/paper_files/paper/2012/file/c399862d3b9d6b76c8436e924a68c45b-Paper.pdf)

## problem
---

- 느린 학습
    - sigmoid, tanh는 입력이 포화 영역에 들어가면 기울기가 작아져 학습이 느련진다.

- 과적합
    - 큰 이미지와 많은 필터를 사용하는 모델을 학습하기 어렵다.

- 복잡한 이미지
    - 좀 더 깊은 이미지 특징 표현을 원했다.

- 계산량
    - 큰 모델을 GPU 하나에 올리기가 어렵다.

**포화 영역** : 입력을 바꿔도 출력이 거의 변하지 않음 (예시 sigmoid에서 0 또는 1에 가까워지는 지점들)

| 입력 \(z\) | sigmoid 출력 | 기울기 |
|---:|---:|---:|

| 5 | 약 0.9933 | 약 0.00665 |
| 10 | 약 0.99995 | 약 0.000045 |
