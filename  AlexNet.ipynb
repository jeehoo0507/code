## problem
---

- 느린 학습
    - sigmoid, tanh는 입력이 **포화 영역**에 들어가면 기울기가 작아져 학습이 느련진다.

- 과적합
    - 큰 이미지와 많은 필터를 사용하는 모델을 학습하기 어렵다.

- 복잡한 이미지
    - 좀 더 깊은 이미지 특징 표현을 원했다.

- 계산량
    - 큰 모델을 GPU 하나에 올리기가 어렵다.

### 1. ReLU 

<div align="center">
  <img src="img/ReLU.png" width="500">
</div>

$$ReLU(z) = max(0, z)$$

- 기존의 문제점 ( ex: sigmoid)

$$\sigma (z) = \frac{1}{1+e^{-z}}$$

신경망에서 loss를 전달할 때 :
- SLP 면? 단순하게 loss를 backpropagation 할 파라미터에 대해 미분하면 된다. 
    - loss의 최소값은 기울기 부호의 반대에 생기니까

<br>

- MLP 면? 여러 층이 존재하기 때문에 backpropagation 할 때 체인룰을 적용해야 한다.
    - 특정 파라미터 w에 loss를 전달한다고 하면 w가 입력에 사용되는 
    <br>가장 가까운 함수를 미분한 값을 가장 마지막 체인룰의 요소로 포함하게 된다.

$$\sigma'(z) = \sigma(z)(1-\sigma(z))$$

sigmoid의 미분값은 최대 0.25이고 입력의 절댓값이 커지면 거의 0이 된다. <br>
- 그림에서도 양끝으로 갈 수록 기울기가 사라짐을 확인 가능하다.

이러한 활성화 함수를 각 층에 넣게 되면 층이 깊어질수록 뒤에 있는 층에 loss 전파가 힘들 것이다.


<br>


- ReLU 해결 아이디어

$$ReLU(z) = max(0, z)$$

$$
\operatorname{ReLU}'(z)=
\begin{cases}
1 & z>0\\
0 & z<0
\end{cases}
$$

### 해결점과 문제점 <br> 

**해결** <br>
기존 sigmoid 나 tanh 에서 나타나는 양, 음 모두에서 기울기가 0에 수렴하는 포화 문제가 해결있었다. <br>
ReLU는 양수 구간에서 미분값이 1이기에 뒤에서 전달 받은 gradient를 그대로 전달한다. 

| 효과 |
|---|
| 기울기 소실 완화 |
| 학습 속도 증가(기존보다 적은 epoch로 학습 가능) |

<br>

**문제** <br>
그러나 특정 뉴런의 입력이 모든 데이터에서 계속 음수이면 ReLU의 출력, 가울기 모두 0이 된다. <br>
이처럼 기울기를 받지 못해 학습이 특정 뉴런의 학습이 안된다는  **Dying ReLU** 문제가 있다.


<br>


이러한 새로운 문제를 해결하기 위해서 Leaky ReLU가 사용된다.

<div align="center">
  <img src="img/leaky ReLU.jpg" width="600">
</div>

음수 구간에도 작은 기울기를 남겨두는 방식이다.

$$
\operatorname{LeakyReLU}(x)=
\begin{cases}
x, & x \geq 0 \\
\alpha x, & x < 0
\end{cases}
\qquad (0 < \alpha \ll 1)
$$
